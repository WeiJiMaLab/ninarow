```python
from tree_search import IBSTracker
from model_fit import SuccessFrequencyTracker
import numpy as np
%load_ext autoreload
%autoreload 2
```

```python
class Predictor(): 
    def __init__(self, p): 
        self.p = p

    def predict(self): 
        if np.random.rand() < self.p: 
            return 1
        else: 
            return 0
```

```python
p = 0.8
predictor = Predictor(p)
actual_move = 1  # true outcome you're estimating likelihood for
nlls = []
for trial in range(int(1e5)):
    tracker = IBSTracker(repeats = 25)
    while not tracker.done:
        tracker.record(predictor.predict() == actual_move)
    
    nlls.append(tracker.nll)
print(f"{np.mean(nlls):.5f} "
      f"± {np.std(nlls):.5f}\t"
      f"{np.log(p):.5f}")
```
