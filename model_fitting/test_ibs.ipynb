```python
from tree_search import IBSTracker
from model_fit import SuccessFrequencyTracker
import numpy as np
```

```python
class Predictor(): 
    def __init__(self, p): 
        self.p = p

    def predict(self): 
        if np.random.rand() < self.p: 
            return 1
        else: 
            return 0
```

```python
p = 0.912
predictor = Predictor(p)
actual_move = 1  # true outcome you're estimating likelihood for
lls = []
for trial in range(int(1e6)):
    tracker = IBSTracker(expt_factor=1)
    while tracker.success_count < tracker.success_threshold:
        predicted_move = predictor.predict()
        if (predicted_move == actual_move):
            tracker.record_success()            
        else:
            tracker.record_failure()
    
    lls.append(tracker.log_likelihood)
print(-np.mean(lls), np.log(p))
```

```python
import numpy as np

p = 0.8
predictor = Predictor(p)

observed = 1

lls = []

for trial in range(int(1e5)):
    tracker = SuccessFrequencyTracker(expt_factor=1)
    delta_ll = 0

    while True:
        y_sim = predictor.predict()

        if y_sim == observed:
            # delta_ll += tracker.report_trial(True)    # <-- required
            break
        else:
            delta_ll += tracker.report_trial(False)

    lls.append(delta_ll)

print(-np.mean(lls), np.log(p))
```
